```python
function highlightOnlyDuplicateContracts() {
  var sheetName = "VOC정보조회"; // 시트명
  var columnToCheck = 1; // 계약번호가 A열 (1열)

  var ss = SpreadsheetApp.getActiveSpreadsheet();
  var sheet = ss.getSheetByName(sheetName);

  if (!sheet) {
    Logger.log("시트명을 다시 확인하세요.");
    return;
  }

  var range = sheet.getDataRange();
  var values = range.getValues();

  var contractMap = {}; // 계약번호별 행 번호 저장
  var duplicateRows = []; // 중복 계약번호가 있는 행만 저장

  // ✅ 계약번호 기준으로 중복 찾기
  for (var i = 1; i < values.length; i++) {
    var contract = values[i][columnToCheck - 1];
    if (contract && contract.toString().trim() !== "") {
      contract = contract.toString().trim();
      if (contractMap[contract]) {
        contractMap[contract].push(i + 1); // 1-based row 번호
      } else {
        contractMap[contract] = [i + 1];
      }
    }
  }

  // ✅ 중복 계약번호가 있는 행만 모으기
  for (var key in contractMap) {
    if (contractMap[key].length > 1) { // 2개 이상 있으면 중복
      duplicateRows = duplicateRows.concat(contractMap[key]);
    }
  }

  // ✅ 먼저 전체 배경색을 흰색으로 초기화
  var lastRow = sheet.getLastRow();
  var lastColumn = sheet.getLastColumn();
  sheet.getRange(2, 1, lastRow - 1, lastColumn).setBackground(null);

  // ✅ 중복된 행만 노란색 음영 처리
  if (duplicateRows.length > 0) {
    duplicateRows.forEach(function(row) {
      var rowRange = sheet.getRange(row, 1, 1, lastColumn);
      rowRange.setBackground("#FFFF99"); // 연노랑
    });
    Logger.log(duplicateRows.length + "개의 중복된 계약번호 행을 노란색으로 표시했습니다.");
  } else {
    Logger.log("중복된 계약번호가 없습니다.");
  }
}
```

```python
function highlightDuplicateContractsOnly() {
  var sheetName = "VOC정보조회"; // 시트명
  var contractColumn = 1; // 계약번호 A열 (1열)

  var ss = SpreadsheetApp.getActiveSpreadsheet();
  var sheet = ss.getSheetByName(sheetName);
  if (!sheet) {
    Logger.log("시트명이 없습니다.");
    return;
  }

  var dataRange = sheet.getDataRange();
  var values = dataRange.getValues();

  var contractMap = {}; // 계약번호 저장
  var duplicateContracts = {}; // 중복 계약번호 저장

  // ✅ 계약번호 중복 체크
  for (var i = 1; i < values.length; i++) {
    var contract = values[i][contractColumn - 1];
    if (contract && contract.toString().trim() !== "") {
      contract = contract.toString().trim();
      if (contractMap[contract]) {
        duplicateContracts[contract] = true; // 중복이면 기록
      } else {
        contractMap[contract] = true; // 처음 등장
      }
    }
  }

  // ✅ 먼저 전체 배경 초기화 (선택사항, 필요없으면 주석처리)
  var lastRow = sheet.getLastRow();
  var lastColumn = sheet.getLastColumn();
  sheet.getRange(2, 1, lastRow - 1, lastColumn).setBackground(null);

  // ✅ 중복 계약번호가 있는 행만 노란색 처리
  for (var i = 1; i < values.length; i++) {
    var contract = values[i][contractColumn - 1];
    if (contract && contract.toString().trim() !== "") {
      contract = contract.toString().trim();
      if (duplicateContracts[contract]) {
        var rowRange = sheet.getRange(i + 1, 1, 1, lastColumn);
        rowRange.setBackground("#FFFF99"); // 연노랑
      }
    }
  }

  Logger.log("✅ 완료: 계약번호 중복된 행만 노란색 음영 처리했습니다.");
}
```

```python
function highlightDuplicatesByContractInDColumn() {
  var sheetName = "VOC정보조회"; // 시트명
  var contractColumn = 4; // ✅ D열은 4번째 열

  var ss = SpreadsheetApp.getActiveSpreadsheet();
  var sheet = ss.getSheetByName(sheetName);
  if (!sheet) {
    Logger.log("시트명을 다시 확인하세요.");
    return;
  }

  var dataRange = sheet.getDataRange();
  var values = dataRange.getValues();

  var contractCount = {}; // 계약번호별 등장 횟수
  var lastRow = sheet.getLastRow();
  var lastColumn = sheet.getLastColumn();

  // ✅ 계약번호별 등장 횟수 계산 (D열만)
  for (var i = 1; i < values.length; i++) { // i=1부터: 데이터 2번째 줄부터 (헤더 제외)
    var contract = values[i][contractColumn - 1]; // D열 (4번째 열)
    if (contract && contract.toString().trim() !== "") {
      contract = contract.toString().trim();
      if (contractCount[contract]) {
        contractCount[contract]++;
      } else {
        contractCount[contract] = 1;
      }
    }
  }

  // ✅ 먼저 전체 배경색 초기화 (선택사항: 주석처리하면 초기화 안 함)
  sheet.getRange(2, 1, lastRow - 1, lastColumn).setBackground(null);

  // ✅ 중복 계약번호가 있는 행만 노란색 처리
  for (var i = 1; i < values.length; i++) {
    var contract = values[i][contractColumn - 1];
    if (contract && contract.toString().trim() !== "") {
      contract = contract.toString().trim();
      if (contractCount[contract] > 1) { // 등장횟수가 2회 이상이면
        var rowRange = sheet.getRange(i + 1, 1, 1, lastColumn);
        rowRange.setBackground("#FFFF99"); // 연노랑
      }
    }
  }

  Logger.log("✅ 완료: D열(계약번호) 중복된 행만 노란색 음영 처리했습니다.");
}
```

```python
import streamlit as st
import pandas as pd
from io import BytesIO
from fpdf import FPDF
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.application import MIMEApplication
from email.mime.text import MIMEText

st.set_page_config(page_title="VOC 중복 분석기", layout="wide")
st.title("📊 VOC 중복 분석 자동화 시스템")

uploaded_file = st.file_uploader("CSV 또는 Excel 파일 업로드", type=["xlsx", "csv"])

# 조건 필터
voc_types = st.multiselect("VOC유형소 필터",
                           ["이용변경불만", "명의변경", "요금이슈", "해지불만"],
                           default=["이용변경불만", "명의변경"])
exclude_status = st.multiselect("제외할 상태",
                                ["처리완료", "진행중", "대기", "미처리"],
                                default=["처리완료"])

if uploaded_file:
    # 파일 읽기
    if uploaded_file.name.endswith('.csv'):
        df = pd.read_csv(uploaded_file)
    else:
        df = pd.read_excel(uploaded_file)

    df.columns = df.columns.str.strip()
    df['계약번호'] = df['계약번호'].astype(str).str.strip()
    df['VOC유형소'] = df['VOC유형소'].fillna('').str.strip()
    df['접수일시'] = pd.to_datetime(df['접수일시'], errors='coerce')

    # 필터 적용
    voc_pattern = '|'.join(voc_types)
    df = df[df['VOC유형소'].str.contains(voc_pattern, regex=True)]
    df = df[~df['상태'].isin(exclude_status)]

    # 중복 분석
    dup_df = df[df.duplicated('계약번호', keep=False)].copy()
    dup_df.sort_values(by=['계약번호', '접수일시'], ascending=[True, False], inplace=True)
    dup_df['최신여부'] = dup_df.duplicated('계약번호', keep='first').apply(lambda x: '' if x else '최신')
    dup_df.insert(0, 'No', dup_df.groupby('계약번호', sort=False).ngroup() + 1)

    st.subheader("중복 분석 미리보기")
    st.dataframe(dup_df.head(20))

    # 엑셀 생성 함수
    def create_excel(data):
        output = BytesIO()
        with pd.ExcelWriter(output, engine='xlsxwriter') as writer:
            data.to_excel(writer, sheet_name='중복분석', index=False)
            ws = writer.sheets['중복분석']
            wb = writer.book

            yellow = wb.add_format({'bg_color': '#FFFF99'})
            blue_bold = wb.add_format({'bold': True, 'font_color': 'blue'})
            red_border = wb.add_format({'top': 1, 'bottom': 1, 'left': 1, 'right': 1, 'border_color': 'red'})

            for i, row in data.iterrows():
                row_num = i + 1
                if data['계약번호'].duplicated(keep=False).iloc[i]:
                    ws.set_row(row_num, None, yellow)
                if row['최신여부'] == '최신':
                    ws.set_row(row_num, None, blue_bold)
                    ws.conditional_format(f"A{row_num+1}:Z{row_num+1}", {
                        'type': 'no_errors', 'format': red_border
                    })

        return output.getvalue()

    # PDF 생성 함수
    def create_pdf(summary):
        pdf = FPDF()
        pdf.add_page()
        pdf.set_font("Arial", size=12)
        pdf.cell(200, 10, txt="VOC 분석 요약", ln=True, align='C')
        for k, v in summary.items():
            pdf.cell(200, 10, txt=f"{k}: {v}", ln=True)
        pdf_output = BytesIO()
        pdf.output(pdf_output)
        return pdf_output.getvalue()

    # 요약정보 생성
    summary_info = {
        "전체 건수": len(df),
        "중복 건수": len(dup_df),
        "중복 그룹 수": dup_df['No'].nunique()
    }

    excel_bytes = create_excel(dup_df)
    pdf_bytes = create_pdf(summary_info)

    st.download_button("📥 엑셀 다운로드", excel_bytes, file_name="VOC_중복분석.xlsx",
                       mime='application/vnd.openxmlformats-officedocument.spreadsheetml.sheet')
    st.download_button("📥 PDF 요약 다운로드", pdf_bytes, file_name="VOC_분석요약.pdf",
                       mime='application/pdf')

    # 이메일 전송
    st.subheader("📧 이메일 전송")
    user_email = st.text_input("받는 사람 이메일 주소")
    send_email = st.button("📨 이메일로 전송")

    if send_email and user_email:
        msg = MIMEMultipart()
        msg['Subject'] = "VOC 분석 결과"
        msg['From'] = "your_email@gmail.com"
        msg['To'] = user_email

        msg.attach(MIMEText("VOC 분석 결과를 첨부드립니다.", 'plain'))

        excel_part = MIMEApplication(excel_bytes, Name="VOC_중복분석.xlsx")
        excel_part['Content-Disposition'] = 'attachment; filename="VOC_중복분석.xlsx"'
        msg.attach(excel_part)

        pdf_part = MIMEApplication(pdf_bytes, Name="VOC_분석요약.pdf")
        pdf_part['Content-Disposition'] = 'attachment; filename="VOC_분석요약.pdf"'
        msg.attach(pdf_part)

        try:
            with smtplib.SMTP_SSL('smtp.gmail.com', 465) as server:
                server.login("your_email@gmail.com", "앱비밀번호")  # 앱 비밀번호 필요
                server.send_message(msg)
            st.success(f"{user_email} 로 이메일 전송 완료!")
        except Exception as e:
            st.error(f"이메일 전송 실패: {e}")
```
