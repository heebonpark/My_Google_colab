```python
import shutil
import os
from google.colab import drive

# Google Drive 마운트
drive.mount('/content/drive')

# 원본 파일명 (Colab에 업로드된 파일명)
file_name = "LOCALDATA_NOWMON_CSV.zip"  # 실제 파일명으로 수정

# 대상 경로 (Google Drive 내 이동하고자 하는 폴더 경로)
target_folder = "/content/drive/My Drive/시설/영업기회"
target_path = os.path.join(target_folder, file_name)

# 대상 폴더가 없으면 생성
if not os.path.exists(target_folder):
    os.makedirs(target_folder)

# 파일 이동
shutil.move(file_name, target_path)

print(f"파일이 구글 드라이브에 업로드되었습니다: {target_path}")
```
