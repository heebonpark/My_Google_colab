```python
# ✅ 라이브러리 로드
import pandas as pd
from google.colab import drive

# ✅ 구글 드라이브 마운트
drive.mount('/content/drive')

# ✅ 파일 경로 설정
file_path = '/content/drive/MyDrive/시설/유지/8월마감 조건추출 결합_선택컬럼_0831.csv'

# ✅ CSV 불러오기 (인코딩 자동 처리)
try:
    df = pd.read_csv(file_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(file_path, encoding='cp949')

# ✅ 관리본부명 치환
df['관리본부명'] = df['관리본부명'].replace({
    '강원본부': '강북/강원본부',
    '서부본부': '강남/서부본부'
})


# ✅ 제외사유가 공백(또는 결측치)인 행만 필터링
df = df[df['제외사유'].isna() | (df['제외사유'].astype(str).str.strip() == '')]

# ✅ KTT월정료 컬럼 전처리 (숫자 변환)
df['KTT월정료(조정)'] = (
    df['KTT월정료(조정)']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.extract('(\d+)')
    .fillna(0)
    .astype(int)
)

# ✅ 그룹별 집계: 관리본부명 기준
summary = df.groupby('관리본부명').agg(
    계약건수=('계약번호', 'nunique'),
    월정료합계백만원=('KTT월정료(조정)', lambda x: round(x.sum() / 1_000_000, 2))  # 백만 원 단위로 변환
).reset_index()

# ✅ 결과 출력
import pandas as pd
from IPython.display import display
pd.options.display.float_format = '{:,.2f}'.format

print("📊 관리본부별 계약 건수 및 월정료 합계 (백만 원 단위):")
display(summary)
```

```python
# ✅ 라이브러리 불러오기
import pandas as pd
from google.colab import drive
from IPython.display import display

# ✅ 구글 드라이브 마운트
drive.mount('/content/drive')

# ✅ 파일 경로
file_path = '/content/drive/MyDrive/시설/유지/8월마감 조건추출 결합_선택컬럼_0831.csv'

# ✅ CSV 불러오기
try:
    df = pd.read_csv(file_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(file_path, encoding='cp949')

# ✅ 관리본부명 치환
df['관리본부명'] = df['관리본부명'].replace({
    '강원본부': '강북/강원본부',
    '서부본부': '강남/서부본부'
})

# ✅ 춘천고객지원팀 → 원주지사로 통합
df['관리지사명'] = df['관리지사명'].replace({'춘천고객지원팀': '원주지사'})

# ✅ 분석 대상 지사만 필터링
target_branches = ['중앙지사', '강북지사', '서대문지사', '고양지사',
                   '의정부지사', '남양주지사', '강릉지사', '원주지사']
df = df[df['관리지사명'].isin(target_branches)]

# ✅ '지사' 제거
df['관리지사명'] = df['관리지사명'].str.replace('지사', '', regex=False)

# ✅ 제외사유 공백 또는 결측치만 필터링
df = df[df['제외사유'].isna() | (df['제외사유'].astype(str).str.strip() == '')]

# ✅ 월정료 숫자 변환
df['KTT월정료(조정)'] = (
    df['KTT월정료(조정)']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.extract('(\d+)')
    .fillna(0)
    .astype(int)
)

# ✅ 강북/강원본부만 필터링
df_branch = df[df['관리본부명'] == '강북/강원본부'].copy()

# ✅ 계약건수: 시설구분 == 대상
fac_target = df_branch[df_branch['시설구분'] == '대상']

# ✅ 월정료: 요금구분 == 대상
fee_target = df_branch[df_branch['요금구분'] == '대상']

# ✅ 계약건수 집계
fac_summary = fac_target.groupby('관리지사명')['계약번호'].nunique().reset_index()
fac_summary.rename(columns={'계약번호': '계약건수'}, inplace=True)

# ✅ 월정료 합계 집계
fee_summary = fee_target.groupby('관리지사명')['KTT월정료(조정)'].sum().reset_index()
fee_summary.rename(columns={'KTT월정료(조정)': '월정료합계원'}, inplace=True)

# ✅ 병합
summary = pd.merge(fac_summary, fee_summary, on='관리지사명', how='outer').fillna(0)

# ✅ 백만원 단위 변환
summary['월정료합계백만원'] = (summary['월정료합계원'] // 1_000_000).astype(int)
summary = summary.drop(columns=['월정료합계원'])

# ✅ 전체합계 추가
total_row = pd.DataFrame({
    '관리지사명': ['전체합계'],
    '계약건수': [summary['계약건수'].sum()],
    '월정료합계백만원': [summary['월정료합계백만원'].sum()]
})
summary = pd.concat([summary, total_row], ignore_index=True)

# ✅ 원하는 지사 순서 적용
branch_order = ['중앙', '강북', '서대문', '고양', '의정부', '남양주', '강릉', '원주']
summary['관리지사명'] = pd.Categorical(summary['관리지사명'], categories=branch_order + ['전체합계'], ordered=True)

# ✅ 순서대로 정렬
summary = summary.sort_values('관리지사명').reset_index(drop=True)

# ✅ 포맷 적용 (천단위 콤마)
summary['계약건수'] = summary['계약건수'].astype(int).map('{:,}'.format)
summary['월정료합계백만원'] = summary['월정료합계백만원'].astype(int).map('{:,}'.format)

# ✅ 최종 출력
print("📋 강북/강원본부 관리지사별 계약건수 및 월정료 (정렬 완료, 단위: 백만원, 천단위 콤마):")
display(summary)
```
